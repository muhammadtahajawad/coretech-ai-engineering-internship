```python
def recommend_service(industry: str, budget: int, urgency: str) -> str:
    """
    Recommends a CoreTech service based on client profile.

    Args:
        industry (str): Client's industry
        budget (int): Client's budget in USD
        urgency (str): 'low', 'medium', or 'high'

    Returns:
        str: Recommended service with reasoning
    """
    recommendation = ""
    reason = ""

    # Budget-based filtering
    if budget < 1000:
        recommendation = "Data Analytics & Reporting"
        reason = "Cost-effective entry point for AI adoption"
    elif budget < 5000:
        recommendation = "AI Chatbot Development"
        reason = "High ROI solution within mid-range budget"
    else:
        recommendation = "Machine Learning Solutions"
        reason = "Full-scale ML pipeline for maximum impact"

    # Industry-specific override
    if industry.lower() in ["manufacturing", "logistics"]:
        recommendation = "AI-Powered Automation"
        reason = "Automation delivers highest value for operational industries"
    elif industry.lower() in ["healthcare", "pharma"]:
        recommendation = "Natural Language Processing"
        reason = "NLP is ideal for medical records and patient communication"
    elif industry.lower() in ["retail", "ecommerce"]:
        recommendation = "AI Chatbot Development"
        reason = "Chatbots directly improve customer experience in retail"

    # Urgency note
    urgency_note = {
        "high": "Prioritized — fast deployment track available",
        "medium": "Standard timeline applies",
        "low": "Flexible scheduling, thorough planning phase"
    }.get(urgency.lower(), "Timeline to be discussed")

    return (
        f"Recommended Service : {recommendation}\n"
        f"Reason              : {reason}\n"
        f"Urgency Note        : {urgency_note}"
    )


# Test the function with multiple client profiles
print("=== Service Recommendations ===")

print("\n[Client 1 — Ahmed Industries]")
print(recommend_service("Manufacturing", 5000, "high"))

print("\n[Client 2 — City Hospital]")
print(recommend_service("Healthcare", 8000, "medium"))

print("\n[Client 3 — Startup]")
print(recommend_service("Tech", 800, "low"))
```

```python
def coretech_assistant():
    """
    A simple command-line assistant for CoreTech Innovation.
    Handles service info, recommendations, and client inquiries.
    """
    print("============================================")
    print("   Welcome to CoreTech Innovation Assistant ")
    print("============================================")
    print("Type 'help' to see available commands.")
    print("Type 'quit' to exit.\n")

    commands = {
        "services" : "List all CoreTech services",
        "recommend": "Get a service recommendation",
        "contact"  : "Get contact information",
        "about"    : "Learn about CoreTech Innovation",
        "help"     : "Show available commands",
        "quit"     : "Exit the assistant"
    }

    while True:
        user_input = input("You: ").strip().lower()

        if user_input == "quit":
            print("Assistant: Thank you for contacting CoreTech Innovation. Goodbye!")
            break

        elif user_input == "help":
            print("\nAssistant: Available commands:")
            for cmd, desc in commands.items():
                print(f"  {cmd:<12} — {desc}")
            print()

        elif user_input == "services":
            print("\nAssistant: Here are our services:")
            for i, s in enumerate(coretech_services, 1):
                print(f"  {i}. {s}")
            print()

        elif user_input == "recommend":
            print("Assistant: Let me help you find the right service.")
            industry = input("  Your industry: ")
            budget = int(input("  Your budget (USD): "))
            urgency = input("  Urgency (low/medium/high): ")
            print("\nAssistant:")
            print(recommend_service(industry, budget, urgency))
            print()

        elif user_input == "contact":
            print("\nAssistant: You can reach CoreTech Innovation at:")
            print("  Email   : info@coretech.io")
            print("  Website : www.coretech.io")
            print("  Hours   : Mon–Fri, 9am–6pm PKT\n")

        elif user_input == "about":
            print("\nAssistant: CoreTech Innovation is an AI solutions company")
            print("  helping businesses adopt intelligent technology to automate,")
            print("  analyze, and grow. Founded in 2018, we serve clients across")
            print("  manufacturing, healthcare, retail, and more.\n")

        else:
            print(f"Assistant: Sorry, I didn't understand '{user_input}'. Type 'help' to see available commands.\n")


# Run the assistant
coretech_assistant()
```
