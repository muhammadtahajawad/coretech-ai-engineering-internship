```python
import pandas as pd
import re

def clean_and_tokenize(text):
    """Normalizes incoming text strings for baseline token extraction."""
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    return set(text.split())

def coretech_bot():
    # Load dataset rules securely
    try:
        data = pd.read_csv("coretech_faq.csv")
    except FileNotFoundError:
        print("Error: 'coretech_faq.csv' not found. Please execute Cell 1 first.")
        return

    print("[CoreTech Automation Agent Active]")
    print("Welcome! I am your rule-based assistant for CoreTech Innovations")
    print("Ask me anything about our services, pricing, or infrastructure pipelines. (Type 'exit' to quit)\n")

    while True:
        user_raw = input("You: ").strip()
        if not user_raw:
            continue

        if user_raw.lower() in ['exit', 'quit', 'bye']:
            print("Chatbot: Thank you for contacting CoreTech Innovations. Have an excellent day!")
            break

        user_tokens = clean_and_tokenize(user_raw)

        # Keyword intersection scoring algorithm
        highest_score = 0
        best_match_idx = -1

        for idx, row in data.iterrows():
            row_keywords = set(str(row['keywords']).split())
            # Calculate overlapping target terms
            score = len(user_tokens.intersection(row_keywords))

            if score > highest_score:
                highest_score = score
                best_match_idx = idx

        # Operational routing evaluation layer
        if highest_score > 0 and best_match_idx != -1:
            reply = data.iloc[best_match_idx]['answer']
            print(f"Chatbot: {reply}\n")
        else:
            print("Chatbot: I apologize, but I cannot locate a specific service mapping for your request. "
                  "Please explore our technical catalogs at coretechio.com or request a formal alignment call.\n")

# Initialize execution trace
coretech_bot()
```
